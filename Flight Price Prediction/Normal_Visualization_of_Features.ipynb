````python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('C:/Users/Admin/Downloads/Clean_Dataset.csv/Clean_Dataset.csv')
df = df.drop(columns=['Unnamed: 0',])

df.head(10)
````

```
Distribution of numerical features.

````python
import seaborn as sns
````

````python
num_cols = df.select_dtypes(include=['int64', 'float64']).columns.tolist()

for col in num_cols:
    sns.histplot(data=df, x=col, kde=True, bins=50, color='skyblue')
    plt.title(f'Distribution of {col}')
    plt.show()
````

````python
for col in num_cols:
    sns.violinplot(data=df, y=col)
    plt.title(f'Outliers of {col}')
    plt.show()
````

```
Price Distribution Accross Categories

````python
df = pd.read_csv('C:/Users/Admin/Downloads/Clean_Dataset.csv/Clean_Dataset.csv')

df['late_booking'] = df['days_left'].apply(lambda x: 'late' if x <= 2 else 'early')
df['route'] = df['source_city'] + ' to ' + df['destination_city']

df = df.drop(columns=['Unnamed: 0', 'days_left', 'source_city', 'destination_city', 'flight'])

cat_cols = df.select_dtypes(include=['object']).columns.tolist()

for col in cat_cols:
    sns.boxplot(data=df, x=col, y='price')
    plt.title(f'{col} vs Price')
    plt.xticks(rotation=90)
    plt.show()
    print()
````

```
Relationship between numerical variables.

````python
num_cols = df.select_dtypes(include=['int64', 'float64']).drop('price', axis=1).columns.tolist()

for col in num_cols:
    sns.scatterplot(data=df, x=col, y='price')
    plt.title(f'{col} vs Price')
    plt.show()
````

```
Counting Frequency of categories

````python
for col in cat_cols:
    sns.countplot(data=df, x=col, color='crimson')
    plt.title(f'Count plot of {col}')
    plt.xticks(rotation=90)
    plt.show()
    print()
````
