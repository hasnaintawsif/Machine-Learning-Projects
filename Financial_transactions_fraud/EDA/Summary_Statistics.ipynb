````python
import pandas as pd
import numpy as np
from scipy import stats
df = pd.read_csv('C:/Users/Admin/Downloads/transactions.csv')
df.head()
````

```
Analyzing mean...

````python
num_cols = [col for col in df.select_dtypes(['int64', 'float64']).columns.tolist() if col != 'isFraud']
for col in num_cols:
    print(f'\nMean of {col}: {np.mean(df[col])}')
````

.

Balances are generally large, suggesting corporate or high-value accounts.

amount is smaller than the difference between old and new balances in some cases, which might indicate inconsistencies or fraud.

newbalanceOrig > oldbalanceOrg on average → unusual. This could highlight reversed transactions, top-ups, or dataset anomalies.

The dataset is likely skewed: a few very high balances or amounts inflate the mean, so median/percentiles are worth checking.

```
Analyzing median...

````python
num_cols = [col for col in df.select_dtypes(['int64', 'float64']).columns.tolist() if col != 'isFraud']
for col in num_cols:
    print(f'\nMedian of {col}: {np.median(df[col])}')
````

.

Data is heavily skewed in all balance and amount columns.

Median is much smaller than mean, so most transactions are small-value, while a few large-value transactions dominate averages.

New balance of origin accounts often ends up 0, showing full depletion is common.

For fraud detection or outlier analysis, focusing on median, quartiles, and extreme percentiles is better than just mean.

```
Analyzing mode...

````python
num_cols = [col for col in df.select_dtypes(['int64', 'float64']).columns.tolist() if col != 'isFraud']
for col in num_cols:
    print(f'\nMode of {col}: {stats.mode(df[col])[0]}')
````

.

Highly skewed and irregular distribution: mean, median, and mode are all very different.

Mode being 0 in balances is a strong sign that many accounts are empty or zeroed out, which is unusual for normal banking data.

Large-value repeated transactions (amount = 10M) could indicate batch processing or fraud patterns.

This confirms your dataset is not normally distributed, with lots of outliers and zeros.

```
Analyzing variance...

````python
num_cols = [col for col in df.select_dtypes(['int64', 'float64']).columns.tolist() if col != 'isFraud']
for col in num_cols:
    print(f'\nVariance of {col}: {np.var(df[col])}')
````

.

Balances and amounts are extremely variable → presence of outliers and high-value transactions dominates the mean.

step variance is much smaller relative to balances, suggesting time is more evenly distributed.

Origin and destination balances have similar magnitude of spread, consistent with skewed transaction behavior (many small accounts, few very large accounts).

High variance + median << mean + mode = 0 → classic skewed, heavy-tailed distribution, possibly due to fraudulent transactions or batch transfers.



```
Analyzing standard_deviation...

````python
num_cols = [col for col in df.select_dtypes(['int64', 'float64']).columns.tolist() if col != 'isFraud']
for col in num_cols:
    print(f'\nStandard_deviation of {col}: {np.std(df[col])}')
````

.

Amounts and balances are heavily skewed, as shown by:

Median << Mean

Mode = 0

High standard deviation

Most transactions are small, while a few are extremely large, which is typical in financial datasets with corporate or fraudulent transactions.

Step (time) is moderately distributed, showing no extreme clustering over time, except perhaps some modes (like step 19).

```
Anayzing minimum, maximum and range...

````python
num_cols = [col for col in df.select_dtypes(['int64', 'float64']).columns.tolist() if col != 'isFraud']
for col in num_cols:
    min_ = np.min(df[col])
    max_ = np.max(df[col])
    ranges = max_ - min_
    print(f'\nMin of {col}: {min_}')
    print(f'Max of {col}: {max_}')
    print(f'Range of {col}: {ranges}')
````

.

Data is extremely skewed and heavy-tailed:

Median << Mean << Max in most columns.

Many zeros (mode = 0) coexist with huge maximums.

Origin accounts often end at zero, recipient accounts can reach very high balances.

Outliers dominate analysis:

Means and standard deviations are inflated by a small number of very large transactions.

Median and quartiles give a better sense of typical transactions.

```
Analyzing percentiles...

````python
num_cols = [col for col in df.select_dtypes(['int64', 'float64']).columns.tolist() if col != 'isFraud']
for col in num_cols:
    q1, q2, q3 = np.quantile(df[col], [.25, .50, .75])
    print(f'\n25th percentile: {q1}, 50th percentile: {q2}, 75th percentile: {q3}')
````

.

Most transactions are small: median << mean.

Many accounts start or end at 0, confirming the heavy presence of empty or new accounts.

Data is extremely skewed: the top quartile contains very large transactions and balances, influencing mean and standard deviation.

Interquartile range (IQR = Q3 – Q1): good measure of typical spread, not influenced by outliers.
