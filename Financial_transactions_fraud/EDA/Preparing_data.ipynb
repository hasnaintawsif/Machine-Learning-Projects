````python
import pandas as pd
df = pd.read_csv('C:/Users/Admin/Downloads/transactions.csv')
df.head()
````

.
Creating more features from existing features including:
```
- How much money left the sender?
- Amount actually received.
- transaction size relative to account balance.
- whether accounts are depleted after the transaction 

````python
df['net_change_org'] = df['newbalanceOrig'] - df['oldbalanceOrg']
df['net_change_dest'] = df['newbalanceDest'] - df['oldbalanceDest']
df['ratio_amount_orig'] = df['amount'] / (df['oldbalanceOrg'] + 1e-5)
df['ratio_amount_dest'] = df['amount'] / (df['oldbalanceDest'] + 1e-5)
df['orig_zero_after'] = (df['newbalanceOrig'] == 0).astype(int)
df['dest_zero_after'] = (df['newbalanceDest'] == 0).astype(int)
````
