**Today’s course focuses on univariate analysis using different visualization and summary techniques. We will perform univariate analysis using:**
- histogram, 
- boxplot,
- violin plot,
- Count plot,
- and pie charts **to understand the distribution and characteristics of individual variables.**

```python
#Import csv file.
import pandas as pd

data = pd.read_csv('C:/Users/Admin/Downloads/Amazon.csv')
data.head()
```

****Histogram****
-

```python
#A histogram shows distribution of a numeric variable by grouping values into bins (ranges) and displaying how many observations
#fall into each bin. It tells us shape of the data (e.g., normal, skewed, uniform, etc.), spread of values, central tendency (where 
#values cluster), presense of outliers, and whether data is left skewed or right skewed. We use a histogram during univariate analysis, 
#before modeling, to understand data distribution, to check skewness or outliers, and to decide whether scaling or transformation is
#needed.

#A histogram is a data visualization method and can be created using matplotlib or seaborn.
import matplotlib.pyplot as plt
import seaborn as sns

#Let's visualize the "TotalAmount" column using a histogram:

plt.figure(figsize=(10,6))
ax = sns.histplot(data=data, x='TotalAmount', bins=30, kde=True)
ax.set_xlabel('Total Amount')
ax.set_ylabel('Count')
ax.set_title('Histogram (Total Amount)')
plt.show()

#Bins split the values into fixed ranges and count how many values fall into each range. Each bar you see is one bin.
#KDE draws a smooth curve that shows the overall shape of the data distribution.
#Together they answer: How many values fall where? And what is the over all distribution shape?
```

Note: From previous result, we can see that: 
1. Most values are small, 
2. very few values are large, 
3. The tail streches to the right (right-skewed distribution).

****Boxplot****
-

```python
#A boxplot is a visualization that summarizes a numeric variable using quartiles and outliers. It shows median, spread,
#skewness, and outliers (very clearly). In a boxplot, the bottom of the box shows Q1 (25th percentile), the top of the
#box shows Q3 (75th percentile), the box itself contains the middle 50% of the data, and the line inside the box shows median
#or Q2 (50th percentile). Between Q1 and Q3, we have the so colled Interquartile Range (IQR). The interquartile range (IQR) is 
#the distance between Q1 and Q3. The T-shaped whiskers extend to the smallest and largest values within 1.5 times the IQR from 
#Q1 and Q3; points beyond this range are considered outliers

#Let's visualize the "UnitPrice" column using a boxplot:

plt.figure(figsize=(10,6))
ax = sns.boxplot(data=data, x='UnitPrice')
ax.set_xlabel('UnitPrice')
ax.set_ylabel('Count')
ax.set_title('Boxplot (UnitPrice)')
plt.show()
```

```python
#The boxplot above does not show any outlier, let's try plotting "TotalAmount" to explore outliers in a boxplot.

plt.figure(figsize=(10,6))
ax = sns.boxplot(data=data, x='TotalAmount')
ax.set_xlabel('TotalAmount')
ax.set_ylabel('Count')
ax.set_title('Boxplot (TotalAmount)')
plt.show()
```

****Violin Plot****
-

```python
#Violin plots are quite similar as the boxplots. See below:

plt.figure(figsize=(10,6))
ax = sns.violinplot(data=data, x='TotalAmount', color='skyblue')
ax.set_xlabel('TotalAmount')
ax.set_ylabel('Count')
ax.set_title('Violinplot (TotalAmount)')
plt.show()
```

****Count Plot****
-

```python
#A countplot is a bar plot for categorical data. It shows the number of occurrences (counts) of each category in a dataset.
#Essentially, it answers the question: How many times does each category appear? It helps us to understand the frequency of
#categories

plt.figure(figsize=(10,6))
ax = sns.countplot(data=data, y='City', color='green')
ax.set_xlabel('Count')
ax.set_ylabel('City')
ax.set_title('Countplot (City)')
plt.show()

#The axes have been swapped (x-axis to y-axis) for batter readability since there are many categories.
```

****Pie Chart****
-

```python
#A pie chart is a circular chart that shows the proportion of each category in a dataset. Each slice represents a category.
#The size of the slice correspondents to the percentage or proportion of that category relative to the whole.

#Let's apply the groupby() method first:
grouped_city = data.groupby('Country')['ShippingCost'].mean()
grouped_city = grouped_city.reset_index()

plt.pie(grouped_city['ShippingCost'], labels=grouped_city['Country'], colors=['green', 'orange', 'red', 'yellow', 'aqua'])
plt.title('Average Tax per Country')
plt.show()

print(grouped_city)
```
