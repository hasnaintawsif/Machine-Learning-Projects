***Now we will generate some basic summary statistics from our dataset, so we are going to work with columns that contain numbers. We are not including Product_ID for it does not contain any useful data that can be used to produce any useful information.***

*We will be analyzing mean, median, mode, variance, standard deviation(std), range, percentile, quartile, and interquartile range(IQR).*

```python
#To generate summary statistics, we have to import Numpy and Stats from Scipy
import numpy as np
from scipy import stats

#Let's select the columns that contain useful numbers and drop 'Product_ID':

num_cols = num_cols = data.select_dtypes(['int64', 'float64']).drop('Product_ID', axis=1).columns.tolist()
print('Num Cols: ', num_cols)
```

```python
#Mean is average value of a numeric column. It answers one simple question, and that is: if all vaues were evenly distributed, 
#what would the typical value be? It is calculated by (Sum of all values/Number of values). If a value is extremely high or 
#low (which is an outlier),it directly increases or decreases the sum, thus shifting the mean.

num_cols = num_cols = data.select_dtypes(['int64', 'float64']).drop('Product_ID', axis=1).columns.tolist()

for col in num_cols:
    print(f'Mean ({col}): ',np.mean(data[col]))
```

```python
#Median is the middle value of a numeric column when sorted in order. If the mean is unusually larger or lower than median, then
#potential outliers exist in the dataset that we are working with.

num_cols = num_cols = data.select_dtypes(['int64', 'float64']).drop('Product_ID', axis=1).columns.tolist()

for col in num_cols:
    print(f'Median ({col}): ',np.median(data[col]))
```

```python
#Mode is the particular value that appears most frequently in a column. However mode can be used for both numeric and object/categorical column.

num_cols = num_cols = data.select_dtypes(['int64', 'float64']).drop('Product_ID', axis=1).columns.tolist()

for col in num_cols:
    print(f'Mode ({col}): ',stats.mode(data[col])[0])
```

```python
#Variance is a masure of how spread out the numbers in a numeric column are from the mean. It answers the question of how far do the
#the values typically deviate from the average value.

num_cols = num_cols = data.select_dtypes(['int64', 'float64']).drop('Product_ID', axis=1).columns.tolist()

for col in num_cols:
    print(f'Variance ({col}): ',np.var(data[col]))
```

```python
#Standard deviation is the square root of variance. Same as variance, but the result is back in the original units of the data.

num_cols = num_cols = data.select_dtypes(['int64', 'float64']).drop('Product_ID', axis=1).columns.tolist()

for col in num_cols:
    print(f'Standard deviation ({col}): ',np.std(data[col]))
```

```python
#Range is the difference between the largest value and the smallest value in a dataset. A single extremely high or low value
#can make the range very high, which also suggest the range is sensetive to outliers.

num_cols = num_cols = data.select_dtypes(['int64', 'float64']).drop('Product_ID', axis=1).columns.tolist()

for col in num_cols:
    max_ = np.max(data[col])
    min_ = np.min(data[col])
    range_ = max_ - min_
    print(f'Range ({col}): ',range_)
```

```python
#Percentile is the value below which a certain percentage of the data falls. It divides the numbers in a numeric column into
#100 equal parts. 50th percentile is the value such that 50% of the data are below it.

num_cols = num_cols = data.select_dtypes(['int64', 'float64']).drop('Product_ID', axis=1).columns.tolist()

for col in num_cols:
    print(f'Percentile 60 ({col}): ',np.percentile(data[col],60))
```

```python
#Quartile is the same as percentile, but it divides a dataset into 4 equal parts.

num_cols = num_cols = data.select_dtypes(['int64', 'float64']).drop('Product_ID', axis=1).columns.tolist()

for col in num_cols:
    print(f'Quartile 75 ({col}): ',np.quantile(data[col], .75))
```

```python
#Interquartile range(IQR) is basically the range of the middle 50% of a particular numeric column or data which 
#is "Q2(percentile 75th) - Q1(percentile 25th)"

num_cols = num_cols = data.select_dtypes(['int64', 'float64']).drop('Product_ID', axis=1).columns.tolist()

for col in num_cols:
    print(f'Interquartile Range ({col}): ',stats.iqr(data[col]))
```
