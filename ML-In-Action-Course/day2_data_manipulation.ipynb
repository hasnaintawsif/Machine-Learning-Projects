
****Grouping****
-

```python
#Grouping data serves the purpose of splitting a dataset into subsets based on one or more columns, then applying calculations
#to each group. It answers questions like: What is the average salary per department? How many customers per country? or What
#is the mean score for each class. And after grouping, we can aggregate the subsets using functions like mean(), sum(), count(), 
#min()/max(), and std.

grouped_data = data.groupby('repeat_retailer')['ratio_to_median_purchase_price'].mean()
print(grouped_data)
grouped_data = data.groupby('repeat_retailer')[['distance_from_home', 'distance_from_last_transaction']].mean()
print('\n',grouped_data)
```

****Appending****
-

```python
#Appending means adding new rows to an existing dataset. For example, if our dataset contains 1,000,000 rows and 8 columns, and we
#want to add rows that are related to this dataset, our new dataset must contain exactly the same columns as the existing one. 
#In this case, the new dataset hasy 8 columns. However, when appending, the number of rows in the new dataset can vary.

data1 = pd.read_csv('C:/Users/Admin/Downloads/card_transdata.csv')
data2 = pd.read_csv('C:/Users/Admin/Downloads/card_transdata.csv')

print('data1 shape: ', data1.shape, '\ndata2 shape: ', data2.shape)

#To append data2 to data1, we will use pd.concat() function:

appended = pd.concat([data1, data2], axis=0)
print('Shape of the appended data: ', appended.shape)

#When appending, writing 'axis=0' is not mandatory, as it is default
```

****Concatenating****
-

```python
#Concatenating is similar to appending, but now instead of adding new rows, we can add new columns to an existing dataset by
#specifying 'axis=1'. See below:

concatenated = pd.concat([data1, data2], axis=1)
print('Shape of the concatenated data: ', concatenated.shape)
```

****Merging****
-

```python
#Merging means combing two datasets based on one or more common columns (keys). Unlike append/concat, which just stack data,
#merge aligns rows based on a key column. There are various types of merge: 1. Inner join keeps only rows with keys present
#in both datasets, 2. Left join keeps all rows from the left dataset, add matching rows from the right, 3. Right join keeps
#all rows from the right dataset, add matching rows from the left, 4. Outer join keeps all rows from both datasets, fill missing
#values with NaN. However, merging defaults to 'Inner join'.

data1 = pd.read_csv('C:/Users/Admin/Downloads/card_transdata_1.csv')
data2 = pd.read_csv('C:/Users/Admin/Downloads/card_transdata_2.csv')
print('Shape of data1: ', data1.shape)
print('Shape of data2: ', data2.shape)

#Inner Join
merged_inner = pd.merge(data1, data2, on='Customer ID', how='inner')
print('\nShape of merged (Inner join) data: ', merged_inner.shape)

#Left Join
merged_left = pd.merge(data1, data2, on='Customer ID', how='left')
print('Shape of merged (Left join) data: ', merged_left.shape)

#Right Join
merged_right = pd.merge(data1, data2, on='Customer ID', how='right')
print('Shape of merged (Right join) data: ', merged_right.shape)

#Outer Join
merged_outer = pd.merge(data1, data2, on='Customer ID', how='outer')
print('Shape of merged (Right join) data: ', merged_outer.shape)

#I just added a column called "Customer ID" manually. See below the datasets I used for merging.
```

****Sorting****
-

```python
#Sorting means arranging the dataset in a specific order, either ascending (small to large) or descending (large to small). Sorting
#can be applied to one or multiple columns.

data = pd.read_csv('C:/Users/Admin/Downloads/card_transdata.csv')

#Sort by Ascending:
data_sorted = data.sort_values(by=['distance_from_home', 'ratio_to_median_purchase_price'], ascending=True)
data_sorted.head(8)

#If you use 'ascending=False', the selected column(s) with descending order. You can also specify the order for multiple columns 
#individually, e.g.,'Ascending=['True', False]' where the first column is sorted ascending and the second descending.
```

****Categorizing****
-

```python
#Categorizing means converting data into categories or groups. It turns data into meaningful buckets or group, making analysis
#and ML easier. When we talk about categorizing data, we specifically refer to binning or cutting a dataset. Here, we will use
#pd.cut() to categorize our numeric data.

#Categorizing "distance_from_home"

import numpy as np
data['categorized_dfh'] = pd.cut(data['distance_from_home'], bins=[0, 500, 4500, np.inf], labels=['low', 'medium', 'high'])

data.head(15)

#The 'categorized_dfh' has been added to the last column, see the result:
```

```python
#You can let pd.cut() function to calculate the bin ranges automatically, but in this case the widths of the ranges may differ, see below:

data['categorized_dfh'] = pd.cut(data['distance_from_home'], bins=3, labels=['low', 'medium', 'high'])
data.head(8)
```

****Removing Duplicate Rows****
-

```python
#Removing duplicate rows means removing rows that are exactly the same or have same values in specific columns. Duplicate data/rows can 
#inflate counts, bias analysis, and affect ML models. Removing duplicates ensures clean, accurate datasets.

#Let's check if any duplicate data exists:
print(data.duplicated().sum())
```

```python
#Let's dive into fillna():

# 1. Fill the empty cells with 0:
data = pd.read_csv('C:/Users/Admin/Downloads/card_transdata(in).csv')

data = data.fillna(0)

data.head(10)
```

```python
# 2.Fill the empty cells with mean()
data = pd.read_csv('C:/Users/Admin/Downloads/card_transdata(in).csv')

data['distance_from_home'] = data['distance_from_home'].fillna(data['distance_from_home'].mean())
data['distance_from_last_transaction'] = data['distance_from_last_transaction'].fillna(data['distance_from_last_transaction'].mean())
data['ratio_to_median_purchase_price'] = data['ratio_to_median_purchase_price'].fillna(data['ratio_to_median_purchase_price'].mean())

data.head(10)
```
